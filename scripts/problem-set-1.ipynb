# Problem Set 1 — SNP File Analysis

This notebook loads the `.snp` file generated in Problem 1 and uses utility functions from `dna_ops.py` to:
1. Clean and compute codon counts for the 3rd line of the file.
2. Clean and compute CG content for the first 10 lines of the file.

```python
import sys

# Make sure Python can find scripts/dna_ops.py
sys.path.append("scripts")

from dna_ops import clean_dna_line, count_codons, cg_content

# Path to .snp file 
snp_path = "data/mouse/chr2.Build37.snp"

with open(snp_path, "r") as f:
    lines = f.readlines()

len(lines)
```

```python
third_line_raw = lines[2]                
third_line_clean = clean_dna_line(third_line_raw)

codon_counts = count_codons(third_line_clean)
codon_counts
```

```python
for i in range(10):
    cleaned = clean_dna_line(lines[i])
    cg = cg_content(cleaned)
    print(f"Line {i+1}: CG content = {cg:.4f}")
```

```python
print(f"Line {i+1}: CG content = {cg*100:.2f}%")
```
